#### Exercício 1

Escreva uma função que receba os valores de três lados de um triângulo e o classifique como "Equilátero" (todos os lados iguais), "Isósceles" (dois lados iguais) ou "Escaleno" (todos os lados diferentes).

```python
# Solução

lado1 = 2
lado2 = 2
lado3 = 3

def verificar_triangulo (a,b,c):
    print("Baseado nos lados deste triângulo, ele é do tipo:")
    if a == b and b == c and a == c: # (a == b == c) também serve
        print("Equilátero")
    elif a != b and b != c and a != c: 
        print("Escaleno")
    else:
        print("Isósceles")

verificar_triangulo(lado1, lado2, lado3)
```

#### Exercício 3 

Você recebeu um dicionário com os nomes dos alunos e suas respectivas notas. Escreva uma função que calcula a média da turma e retorna uma lista com os nomes dos alunos que tiveram nota acima da média.

```python
# LIÇÕES APRENDIDAS INSERIDAS EM FORMA DE COMENTÁRIOS AO LONGO DO EXERCÍCIO!
turma1 = {
    "caio": 8.0,
    "fabiano": 4.9,
    "juliana": 7.0,
    "fernanda": 9.8,
    "lemos": 5.9,
    "ventura": 7.0,
    "mariana": 3.2,
    "arthur": 5.0,
    "eduardo": 6.8,
    "maurício": 8.5
}


def alunos_aprovados():
    # --- CALCULANDO A MÉDIA ---

    soma_notas = 0 

    # Sempre que quisermos trabalhar com uma iteração cumulativa, precisamos criar um "balde" vazio primeiro.
    # Este "balde" seria uma variável como a soma_notas acima, INT. Ou como veremos abaixo, uma lista vazia aprovados [].
    # Isso é necessário pois NÃO é possível criar uma variável na mesma linha em que fazemos o comando de "+=", por exemplo.

    for aluno, nota in turma1.items(): 
        soma_notas += nota

    # items permite que trabalhemos tanto com as chaves (keys) quanto com os valores (values) de forma natural.
    # para chamar um ou outro, só utilizar keys ou values. Items permite trabalhar com ambos de uma só vez.

    media = soma_notas / len(turma1)
    print(f"A média da turma é: {media:.2f}") # ":.Xf" serve pra limitar a quantidade de casas decimais exibidas em X casas.

    # --- RETORNANDO OS ALUNOS QUE FICARAM ACIMA DA MÉDIA ---
    aprovados = []
    print("\nOs alunos que ficaram acima da média foram:")
    for aluno, nota in turma1.items(): 
        if nota > media:
            aprovados.append(aluno)
    return aprovados
            

alunos_aprovados()
```

#### Exercício 7

Crie uma função que receba uma lista de números inteiros e retorne um dicionário contendo a contagem de quantos números são pares e quantos são ímpares.

```python
# Solução
lista = list(range(10))

def quantidade_par_impar(lista_numeros):
    par = 0 
    impar = 0
    for x in lista_numeros:
        if x % 2 == 0:
            par += 1 
        else:
            impar += 1

    numeros_organizados = dict(pares = par, impares = impar)
    return numeros_organizados

quantidade_par_impar(lista)


# DESAFIO EXTRA DO GEMINI: Como fazer essa mesma função criar um dicionário com os elementos que foram filtrados?

def lista_par_impar(lista_numeros):
    par = []
    impar = []

    for x in lista_numeros:
        if x % 2 == 0:
            par.append(x)
        else:
            impar.append(x)

    numeros_organizados = dict(pares = par, impares = impar)
    return numeros_organizados

lista_par_impar(lista)
```
